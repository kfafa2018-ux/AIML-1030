Developing A classification System to classify articles into genres like religion, science etc. Uses a 20 newsgroup dataset. This model will be saved to call when necessary. We will use the Naive Bayes ckassifier, and a SVM classifer.

```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

# Load the dataset
newsgroups = fetch_20newsgroups(subset='all', remove=('headers', 'footers', 'quotes'))

# Print dataset keys and description
print(f"Target classes (Genres): {list(newsgroups.target_names)}")
print(f"Total documents: {len(newsgroups.data)}")
```

```python
#importing scikit learn library, the dataset, countVectorizer,
# naive bayes, SGD clssifier, TF-IDF, pipeline.
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import SGDClassifier
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.pipeline import Pipeline
```

```python
#For claculations import numpy
import numpy as np
```

```python
#Reading the file
#subset='train' chooses the training set from the whole data.
#shuffle is used to randomize the order of the training data
trainData = fetch_20newsgroups(subset='train', shuffle=True)
```

The  Pipeline is designed using the Pipeline function from sklearn library.

```python
#Designing the pipeline
classificationPipeline=Pipeline([('bow', CountVectorizer()), ('vector', TfidfTransformer()), ('classifier', MultinomialNB())])
```

The Pipeline consists of 3 stages.


1.   Convert the text into features (done using CountVectorizer)
2.   Calculate the term frequency & the inverse document frequency.(Done using the TfidfTransformer)
1. The final stage. Choosing the classifier. The 2 options discussed were Naive Bayes & SVM.







Naive Bayes Classifiers are one of the simplest models used for classification. They assume that the features of input data are independent of one another.

```python
#Option A: Naive Bayes Classifier
#fit function is used to fit the classifier to data
#('classifier'), MultinomialNB()) is used for the Naive Bayes Classifier
#trainData.data & trainData.target refer to the training data & training labels


#classificationPipeline=classificationPipeline.fit(trainData.data, trainData.target)
```

SVM stands for Support Vector Machine. Are typical classifiers used for segreegating datainto classes. They are a type of supervised learning algorithm. SVMs classify datapoints into one of 2 classes by finding the hyperplane between them and widening the margin.

```python
#Option B: SVM Classifier
# Setting up the pipeline with an SGDClassifier (SVM with hinge loss)
classificationPipeline = Pipeline([
    ('bow', CountVectorizer()),
    ('vector', TfidfTransformer()),
    ('classifier', SGDClassifier(loss='hinge', penalty='l1', alpha=0.0005, l1_ratio=0.17))
])

# Fit the SVM pipeline to the training data
classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
```

```python
#Model Prediction
#Test dataset is chosen by the 'test' value for parameter subset.
#Predict function is used for model prediction.
#Accuracy is given by np.mean(dataPrediction==test_set.target)
test_set=fetch_20newsgroups(subset='test', shuffle=True)
dataPrediction=classificationPipeline.predict(test_set.data)
```

Web Application
Building a web application that will use the model for classifying newsgroup data. It will consist of a simple web page that provides newsgroup classification functions.

```python
#main app file

#import file management libraries
import os
import glob

#import streamlit
import streamlit as st

#import scikit-learn library & important functions
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import SGDClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.pipeline import Pipeline

#import numpy for calculations
import numpy as np

#creating a select box in the sidebar with Naive Bayes
#& SVM classification
cs=["Naive Bayes", "SVM"]
classification_space=st.sidebar.selectbox("Pick a classification method:", cs)

st.write("Results")
st.write('Dataset details here:')
st.write("Twenty Newsgroup dataset chosen. It contains about 18000 different posts from newspapers & 20 different topics.")
#displays informative text on the screen

#The app has a classify button
#Depending on if NB or SVM is selected in the checkbox a particular function will be called

#A button name can be created using st.sidebar.button & we must
#pass the button name . If the button is selected,
#The st.sidebar.button returns true
if st.sidebar.button('Classify'):
    #CLASSIFYING THE DATA
    trainData = fetch_20newsgroups(subset='train', shuffle=True)

    if classification_space == "Naive Bayes":
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', MultinomialNB())
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
        #DEFINING TEST SET
        test_set = fetch_20newsgroups(subset='test', shuffle=True)
        dataPrediction = classificationPipeline.predict(test_set.data)
        st.write("Accuracy of Naive Bayes:")
        #PRINTING ACCURACY
        st.write(np.mean(dataPrediction == test_set.target))

    #SVM Classifier
    elif classification_space == "SVM":
        st.write("SVM selected")
        #DEFINING PIPELINE
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', SGDClassifier(loss='hinge', penalty='l1', alpha=0.0005, l1_ratio=0.17))
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)

        #DEFINING TEST SET
        test_set = fetch_20newsgroups(subset='test', shuffle=True)
        dataPrediction = classificationPipeline.predict(test_set.data)
        st.write("SVM:")
        #PRINTING ACCURACY
        st.write(np.mean(dataPrediction == test_set.target))
```

```python
%%writefile app.py
import streamlit as st
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import SGDClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.pipeline import Pipeline

# Side bar navigation
cs = ["Naive Bayes", "SVM", "Random Forest", "Decision Tree"]
classification_space = st.sidebar.selectbox("Pick a classification method:", cs)

st.write("## Results")
st.write("**Dataset details:**")
st.write("The Twenty Newsgroups dataset contains about 18,000 posts across 20 different topics.")
st.write("This app uses these models for classification: Naive Bayes, SVM, Random Forest, Decision Tree")

if st.sidebar.button('Classify'):
    st.write(f"Running classification using: **{classification_space}**...")

    # Fetch datasets
    trainData = fetch_20newsgroups(subset='train', shuffle=True)
    test_set = fetch_20newsgroups(subset='test', shuffle=True)

    if classification_space == "Naive Bayes":
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', MultinomialNB())
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
        dataPrediction = classificationPipeline.predict(test_set.data)
        accuracy = np.mean(dataPrediction == test_set.target)
        st.success(f"Accuracy of Naive Bayes: {accuracy:.4f}")

    elif classification_space == "SVM":
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', SGDClassifier(loss='hinge', penalty='l1', alpha=0.0005, l1_ratio=0.17))
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
        dataPrediction = classificationPipeline.predict(test_set.data)
        accuracy = np.mean(dataPrediction == test_set.target)
        st.success(f"Accuracy of SVM: {accuracy:.4f}")

    elif classification_space == "Random Forest":
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', RandomForestClassifier(max_depth=2, random_state=0))
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
        dataPrediction = classificationPipeline.predict(test_set.data)
        accuracy = np.mean(dataPrediction == test_set.target)
        st.success(f"Accuracy of Random Forest: {accuracy:.4f}")

    elif classification_space == "Decision Tree":
        classificationPipeline = Pipeline([
            ('bow', CountVectorizer()),
            ('vector', TfidfTransformer()),
            ('classifier', DecisionTreeClassifier())
        ])
        classificationPipeline = classificationPipeline.fit(trainData.data, trainData.target)
        dataPrediction = classificationPipeline.predict(test_set.data)
        accuracy = np.mean(dataPrediction == test_set.target)
        st.success(f"Accuracy of Decision Tree: {accuracy:.4f}")
```
